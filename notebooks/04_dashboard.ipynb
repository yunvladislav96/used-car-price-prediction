```python
import pandas as pd
import numpy as np
import plotly
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Чтобы графики отображались прямо в ноутбуке
import plotly.io as pio
pio.renderers.default = "notebook"

# Чтобы pandas не обрезал колонки при выводе
pd.set_option("display.max_columns", 50)
pd.set_option("display.width", 200)

print("pandas:", pd.__version__)
print("plotly:", plotly.__version__)
```

```python
# Загружаем очищенные данные
df = pd.read_csv("../data/processed/cars_clean.csv")

print("Размер таблицы:", df.shape)
df.head(3)
```

```python
from catboost import CatBoostRegressor

model = CatBoostRegressor()
model.load_model("../models/catboost_model.cbm")

print("Модель загружена")
print("Количество признаков:", len(model.feature_names_))
print("Признаки:")
for i, name in enumerate(model.feature_names_, 1):
    print(f"  {i}. {name}")
```

```python
import plotly.io as pio
pio.renderers.default = "notebook_connected"
print("Рендерер:", pio.renderers.default)
```

```python
# График 1: зависимость цены от года выпуска
fig1 = px.scatter(
    df,
    x="год",
    y="цена",
    color="цена",                    
    color_continuous_scale="Viridis", # палитра
    hover_data={
        "марка": True,
        "модель": True,
        "пробег": ":,",
        "мощность": True,
        "цена": ":,",
        "год": True,
    },
    opacity=0.5,
    title="Цена vs год выпуска (4006 объявлений)",
)

fig1.update_layout(
    xaxis_title="Год выпуска",
    yaxis_title="Цена, руб.",
    height=600,
    coloraxis_showscale=False,       
)

fig1.show()
```
